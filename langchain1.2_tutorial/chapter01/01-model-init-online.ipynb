```python
import os
import uuid

from dotenv import load_dotenv
# from langchain.chat_models import init_chat_model

load_dotenv(override=True)
from langchain_openai import ChatOpenAI

# 读取环境变量（不要加引号）
OPENCODE_GO_API_KEY = os.getenv("OPENCODE_GO_API_KEY")

OPENCODE_GO_API_BASE = os.getenv("OPENCODE_GO_API_BASE")

session_id = str(uuid.uuid4())

llm = ChatOpenAI(
    model="deepseek-v4.1-flash",
    openai_api_base=OPENCODE_GO_API_BASE,   # ✅ 这里是 URL
    openai_api_key=OPENCODE_GO_API_KEY,                # ✅ 这里是变量值
    default_headers={
        "x-opencode-session": session_id,
    },
)

response = llm.invoke("用一句话解释 SOLID 原则。")
print(response.content)
```
